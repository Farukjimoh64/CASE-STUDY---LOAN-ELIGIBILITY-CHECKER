TASK 1 - CREATING THE CUSTOMER VARIABLES 

```python
# To create variables using the input() function

customer_name = input("Please, Enter Name").strip()
customer_id = input("Please, Enter your ID")
customer_age = int(input("Please, Enter your Age"))
monthly_income = int(input("Enter your Monthly Income"))
existing_monthly_debt = int(input("Enter Existing Monthly debt"))
requested_loan_amount = int(input("Enter Loan Amount"))
loan_term = int(input("Enter Loan Term"))
credit_score = int(input("Enter Credit Score"))
existing_loan_status = input("Existing Loan:").strip().title() == "False"
employment_status = "Employed"
```

TASK 2 - VALIDATING CUSTOMER INFORMATION

```python
# To validate customer information
## Testing whether the input are of the right data type

type(customer_name) == str
type(customer_id) == str
type(customer_age) == int
type(monthly_income) == int
type(requested_loan_amount) == int
```

TASK 3 - VALIDATING CUSTOMER AGE 

```python
# To validate the age, check the below conditions; if it is met, it returns "Customer age is valid". Else, it returns otherwise

if 18 <= customer_age < 65:
    print(f"Customer Age is Valid!: {customer_age} years")
else: 
    print("Age is Invalid. Enter the right Age")
```

TASK 4 - VALIDATING MONTHLY INCOME

```python
# To validate monthly income if it meets the requirement. it returns "Passed" if it aligns, but returns "Failed" if otherwise

if monthly_income >= 150000:
    print(f"Income Requirement: PASSED {monthly_income} Naira")
else:
    print("Income Requirement: FAILED")
```

TASK 5 - VALIDATING LOAN AMOUNT

```python
# To validate Loan Amount; the Loan must be greater than 0 AND less than 5000000

if 0 < requested_loan_amount <= 5000000:
    print(f"Valid Loan Amount: {requested_loan_amount} Naira")
else:
    print("Invalid! Try Again")
```

TASK 6 - DETERMINING THE CREDIT SCORE CATEGORY

```python
# To determine the credit score using varying conditions; Excellent, Good, Fair, and Poor

if credit_score >= 750:
    status = "Excellent"
elif credit_score >= 650:
    status = "Good"
elif credit_score >= 550:
    status = "Fair"
else: 
    status = "Poor"
print(f"Credit Score {credit_score} - {status}")
```

TASK 7 - CALCULATING THE DEBT-INCOME-RATIO

```python
# To check if the Debt_income_ratio meets the requirement

debt_income_ratio = (existing_monthly_debt/monthly_income)*100 
print(f"Debt_income_ratio: {debt_income_ratio} %")
if debt_income_ratio <= 40:
    print("Requirement PASSED: Please Proceed")
else:
    print("Requirement FAILED: Try Again")
```

TASK 8 - VALIDATING EMPLOYMENT STATUS

```python
# To check if the employment status is in alignment 

if employment_status == "Employed":
    print("VALID: You can proceed")
else:
    print("INVALID: Sorry! You'd have to withdraw from here")
```

TASK 9 - CHECKING EXISTING LOAN STATUS

```python
# To check if there is an existing loan

if existing_loan_status == "False":
    print("Existing Loan: None")
else:
    print("Existing Loan: Active")
```

TASK 10 - CHECKING LOAN TERM

```python
# To check if Loan term is among the suggested ones; 6,12,or 24

if loan_term in [6,12,24]:
    print(f"ACCEPTED: {loan_term} months")
else:
    print("INVALID Loan Term")
```

TASK 11 - CREATING A LOAN ELIGIBILITY FUNCTION

```python
# To create the Loan eligibility function

def check_loan_eligibilty(
    customer_age,
    monthly_income,
    debt_income_ratio,
    credit_score,
    existing_loan_status,
    employment_status,
    requested_loan_amount,
    loan_term
):     # This is a defined function that stores the parameters for eligibility
    customer_name = input("Please, Enter Name")
    customer_id = input("Please, Enter your ID")
    customer_age = int(input("Please, Enter your Age"))
    monthly_income = int(input("Enter your Monthly Income"))
    existing_monthly_debt = int(input("Enter Existing Monthly debt"))
    requested_loan_amount = int(input("Enter Loan Amount"))
    loan_term = int(input("Enter Loan Term"))
    credit_score = int(input("Enter Credit Score"))
    existing_loan_status = input("Existing Loan:").strip().title() == "False"
    employment_status = "Employed" 

    name = customer_name.strip().title()
    
    valid_age = 18 <= customer_age < 65
    valid_income = monthly_income >= 150000
    valid_credit_score = credit_score >= 650
    valid_debt_ratio = debt_income_ratio <= 40
    no_existing_loan = existing_loan_status == False
    valid_employment = employment_status == "Employed"
    valid_loan_amount = 0 < requested_loan_amount <= 5000000
    valid_loan_term = loan_term in [6,12,24]

    loan_eligibility = (
        valid_age
        and valid_income
        and valid_loan_amount
        and valid_credit_score
        and valid_debt_ratio
        and no_existing_loan
        and valid_employment
        and valid_loan_term
    )
    return loan_eligibility
check_loan_eligibilty(
    customer_age,
    monthly_income,
    debt_income_ratio,
    credit_score,
    existing_loan_status,
    employment_status,
    requested_loan_amount,
    loan_term
)


# it should receive the necesary and evaluate the eligibilty rules, also return the appropriate result
```
